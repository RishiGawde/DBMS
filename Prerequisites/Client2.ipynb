```python
import socket
import time

# Server details
SERVER_IP = "127.0.0.1"   # Change to server IP if on another computer
SERVER_PORT = 5000

HOST = "127.0.0.1"
# Listen on all interfaces
PORT = 6000

# Create UDP socket
object2 = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

# Bind client socket (optional)
object2.bind((HOST, PORT))

message = "Hello Server. I am client 2"

while True:
    object2.sendto(message.encode(), (SERVER_IP, SERVER_PORT))
    data, server = object2.recvfrom(1024)
    print("Server:", data.decode())
    time.sleep(1)

object2.close()
```
